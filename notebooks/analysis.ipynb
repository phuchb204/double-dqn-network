# Phân tích kết quả thực nghiệm
Notebook đọc  (mean±std theo seed) và trình bày bảng so sánh DQN / Double DQN / Dijkstra. Bổ sung nhận xét cho từng mục.

```
import pandas as pd
pd.set_option("display.precision", 3)
summary = pd.read_csv("results/csv/summary.csv")
summary
```

```
def fmt(row, col):
    s = row[f"{col}_std"]
    if pd.isna(s): return f"{row[col]:.3f}"
    return f"{row[col]:.3f} ± {s:.3f}"

out = summary.copy()
out["reward"] = summary.apply(lambda r: fmt(r, "reward"), axis=1)
out["delay_ms"] = summary.apply(lambda r: fmt(r, "delay_ms"), axis=1)
out["loss_rate"] = summary.apply(lambda r: fmt(r, "loss_rate"), axis=1)
out["throughput"] = summary.apply(lambda r: fmt(r, "throughput"), axis=1)
out = out[["load","algo","reward","delay_ms","loss_rate","throughput","n_seeds"]]
out
```

### Nhận xét
- RL vs Dijkstra: RL giảm mất gói ~x%, delay ~y%, throughput tăng ~z%.
- DQN vs Double DQN: độ ổn định (std), tốc độ hội tụ.
