# Phân tích kết quả thực nghiệm

Notebook này đọc các file CSV trong `results/csv/` và trình bày bảng so sánh
cuối cùng giữa DQN, Double DQN và baseline Dijkstra. Bổ sung nhận xét ở các
cell markdown khi có kết quả chính thức.

```
import pandas as pd
pd.set_option('display.precision', 4)
pd.set_option('display.float_format', lambda x: f'{x:.4f}')

summary = pd.read_csv('results/csv/summary.csv')
summary
```

### Nhận xét

- DQN vs Double DQN: độ ổn định, tốc độ hội tụ, reward cuối cùng?
- RL vs Dijkstra: cải thiện loss rate / delay / throughput ở mức tải nào?
