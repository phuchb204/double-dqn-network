# Chạy thực nghiệm Double DQN trên Colab

Notebook này: clone repo -> cài dependencies -> chạy baseline + huấn luyện DQN/Double DQN -> vẽ biểu đồ. Kết quả nằm trong `results/`.

Có thể chỉnh các biến ở cell cấu hình (số episode, mức tải, ...).

````
# === CONFIG ===
REPO_URL = 'https://github.com/hoangbaophuc/double-dqn-network.git'
TOPOLOGY = 'hub'
LOADS = '2,4,6,8'              # baseline chạy trên các mức tải
TRAIN_LOADS = [6.0, 8.0]       # huấn luyện DQN/DDQN trên các mức tải này
EPISODES = 2000                # số episode huấn luyện
EVAL_EVERY = 200
EVAL_EPISODES = 20
LR = 1e-3
BATCH_SIZE = 128
GAMMA = 0.99
EPS_DECAY_STEPS = 30000
````

````
# Cài đặt (chạy 1 lần)
!git clone {REPO_URL} double-dqn-network 2>/dev/null || (cd double-dqn-network && git pull)
%cd double-dqn-network
!pip install -q -r requirements.txt
````

````
# Baseline Dijkstra trên nhiều mức tải
!python experiments/baseline.py --topology {TOPOLOGY} --loads {LOADS} --episodes 50
````

````
for load in TRAIN_LOADS:
    print(f'===== DQN load={load} =====')
    !python experiments/train_dqn.py --topology {TOPOLOGY} --load {load} \
        --episodes {EPISODES} --eval-every {EVAL_EVERY} --eval-episodes {EVAL_EPISODES} \
        --lr {LR} --batch-size {BATCH_SIZE} --gamma {GAMMA} --eps-decay-steps {EPS_DECAY_STEPS}
````

````
for load in TRAIN_LOADS:
    print(f'===== Double DQN load={load} =====')
    !python experiments/train_ddqn.py --topology {TOPOLOGY} --load {load} \
        --episodes {EPISODES} --eval-every {EVAL_EVERY} --eval-episodes {EVAL_EPISODES} \
        --lr {LR} --batch-size {BATCH_SIZE} --gamma {GAMMA} --eps-decay-steps {EPS_DECAY_STEPS}
````
